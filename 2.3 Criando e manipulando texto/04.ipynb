Imutabilidade: novas strings criadas a partir de outras strings

```python
minha_str = 'livro python 3'
# minha_str[-1] = '2' ==> TypeError
minha_str
```

```python
minha_str = minha_str[:-1] + '2'
minha_str
```

```python
minha_str = 'livro python 3'
minha_str = minha_str.replace('3', '2')
minha_str
```
