Condicionais: if, elif & else

```python
from utils.chk_numbers import isnumber
```

```python
salario = input('Salário? ')
imposto = input('Imposto em % (ex: 27.5)? ')
```

```python
if not isnumber(salario):
    salario = 1212
    print('Salário mínimo definido.')
else:
    salario = int(salario)
```

```python
if not isnumber(imposto):
    imposto = 27.5
    print('Imposto padrão definido.')
else:
    imposto = float(imposto)
```

```python
total = salario - salario * (imposto * 0.01)
```

```python
print(f'Valor real: R${total:.2f}')
```
