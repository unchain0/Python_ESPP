```python
from utils.chk_numbers import isnumber
```

```python
salario = input('Salário?').replace(',', '.')
imposto = 27.
```

```python
if isnumber(salario):
    salario = float(salario)

    while True:
        resp = input('Imposto ou (s) para sair')
        if resp.lower()[0] == 's':
            imposto = float(imposto)
            break

        if isnumber(resp):
            imposto = float(resp)
        else:
            imposto = 27.
```

```python
total = salario - salario * (imposto * 0.01)
```

```python
print(f'Valor real: R${total:.2f}')
```
