```python
from utils.chk_numbers import isnumber
```

```python
salario = input('Salário?').replace(',', '.')
imposto = 27.
```

```python
if isnumber(salario):
    salario = float(salario)
    resp = None

    while resp != 0:
        resp = input('Imposto ou (0) para sair:')
        if resp == '0':
            resp = int(resp)
            continue
        else:
            if not isnumber(resp):
                imposto = 27.5
            else:
                imposto = float(resp)

    total = salario - salario * (imposto * 0.01)
    print(f'Valor real: R${total}')
else:
    print('Salário inválido!')
```
