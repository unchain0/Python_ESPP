# Unpacking dos argumentos

```python
def unpacking_experiment(*args):
    arg1 = args[0]
    arg2 = args[1]
    others = args[2:]
    print(f'Primeiros argumento => {arg1}')
    print(f'Segundo argumento => {arg2}')
    print(f'Outros argumentos => {others}')
```

```python
unpacking_experiment(1, 2, 3, 4, 5, 6)
```

```python
def unpacking_experiment(**kwargs):
    for k, v in kwargs.items():
        print(f'{k}: {v}')
```

```python
unpacking_experiment(named='Test', other='Other')
```
