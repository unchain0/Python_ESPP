# Packing

```python
from datetime import date
```

```python
d = (2022, 7, 15)
```

```python
date(*d)
```

```python
def new_user(active=False, admin=False):
    print(active)
    print(admin)
```

```python
config = {'active': False,
          'admin': True}
```

```python
new_user(**config)  # * mostra as chaves, ** mostra os valores
```
