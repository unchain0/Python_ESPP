```python
from utils.chk_numbers import isnumber
```

```python
imposto = input('Imposto: ')
```

```python
if isnumber(imposto):
    imposto = float(imposto)
    if imposto >= 0:
        if imposto < 10.:
            print('Baixo')
        elif imposto <= 27.:
            print('Médio')
        elif imposto < 100:
            print('Alto')
    else:
        print('Imposto inválido!')
else:
    print('Imposto deve ser um número!')
```
